```python
!pip install duckdb
import duckdb

db = duckdb.connect('/Users/leenisa/Documents/GitHub/Astra-Lost-In-Space/backend/fires.duckdb')

# Get all fires
fires = db.execute("SELECT * FROM all_fires").fetchall()
print(fires)

# Get statistics
stats = db.execute("""
    SELECT 
        COUNT(*) as total,

        AVG(confidence) as avg_confidence,
        MAX(brightness) as max_brightness
    FROM all_fires
""").fetchall()
print(stats)

db.close()
```
